## Chart 1: Misleading Correlation Using Dual Y-Axes

### Original Chart

![Original Chart](flaw2.webp)

### Design Flaws
1. Dual Y-Axes Create False Correlation

The chart uses two different vertical scales:

Left axis: Units Sold (100–150)
Right axis: Stock Price (10–30)

Because the scales are independently adjusted, the lines appear to move together even if the relationship is weak.

2. Different Units Are Compared Directly

Units sold and stock price measure completely different things.

Displaying them on the same plot encourages viewers to visually compare values that are not directly comparable.

3. Correlation Cannot Be Accurately Judged

The chart visually suggests a strong positive relationship.

However, the apparent similarity is largely created by axis manipulation rather than actual statistical evidence.

A scatter plot or normalized comparison would communicate the relationship more honestly.

### Redesign Fixes Applied

✓ Removed secondary y-axis

✓ Converted both metrics into percentage of their first value

✓ Used a single y-axis

✓ Added clear labels and legend

```python
import pandas as pd
import matplotlib.pyplot as plt

months = ['Jan','Feb','Mar','Apr','May','Jun']

units_sold = [100,120,130,125,140,150]
stock_price = [10,15,30,25,28,27]

df = pd.DataFrame({
    'Month': months,
    'Units Sold': units_sold,
    'Stock Price': stock_price
})

# Normalize to January = 100
df['Units Index'] = df['Units Sold']/df['Units Sold'][0]*100
df['Price Index'] = df['Stock Price']/df['Stock Price'][0]*100

plt.figure(figsize=(8,5))

plt.plot(df['Month'], df['Units Index'],
         marker='o', label='Units Sold Index')

plt.plot(df['Month'], df['Price Index'],
         marker='s', label='Stock Price Index')

plt.title('Normalized Comparison of Sales and Stock Price')
plt.ylabel('Index (Jan = 100)')
plt.xlabel('Month')
plt.legend()
plt.grid(True)

plt.show()
```

| Before                         | After                        |
| ------------------------------ | ---------------------------- |
| Dual y-axes                    | Single y-axis                |
| False visual correlation       | Honest comparison            |
| Different scales               | Common scale                 |
| Hard to interpret relationship | Easy to compare growth rates |


### Explanation

The redesign converts both variables to an index with January = 100. This allows viewers to compare relative change rather than absolute values, eliminating the misleading effect created by independent y-axis scales.

## Chart 2: Florida Gun Deaths (Reuters)
### Original Chart

![Original Chart](flaw1.webp)

### Design Flaws
1. Inverted Y-Axis

The chart places:

0 at the top
1000 at the bottom

This is opposite to how most line charts are interpreted.

Viewers may incorrectly perceive decreases as increases.

2. Large Filled Area Creates Visual Distortion

The red background and gray filled region dominate the graphic.

The area attracts more attention than the actual trend.

This exaggerates perceived changes.

3. Political Annotation Bias

The chart highlights the "Stand Your Ground" law directly on the graph.

The annotation can encourage viewers to infer causation despite the chart only showing correlation.

### Redesign Fixes Applied

✓ Standard y-axis orientation

✓ Removed unnecessary area fill

✓ Used a clean line chart

✓ Added proper axis labels

✓ Kept annotation separate from trend interpretation

```python
import matplotlib.pyplot as plt

years = list(range(1988, 2013))

murders = [
    873,800,790,800,730,680,670,630,590,450,
    500,500,550,590,550,520,730,820,780,690,
    660,690,730,721,710
]

plt.figure(figsize=(10,5))

plt.plot(years, murders, marker='o')

plt.title('Firearm Murders in Florida')
plt.xlabel('Year')
plt.ylabel('Number of Murders')

plt.grid(True)

plt.show()
```

| Before                | After                  |
| --------------------- | ---------------------- |
| Inverted axis         | Standard axis          |
| Heavy area fill       | Clean line chart       |
| Visual exaggeration   | Accurate trend display |
| Suggestive annotation | Neutral presentation   |


### Explanation

The redesign follows standard chart conventions by placing larger values higher on the y-axis. Removing the area fill reduces visual clutter and prevents exaggeration of fluctuations. The result focuses attention on the actual trend rather than graphical effects.
